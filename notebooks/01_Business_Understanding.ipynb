# Customer Churn Prediction & Business Insights

## Business Problem

Customer churn is one of the biggest challenges faced by telecom companies.
Acquiring a new customer is significantly more expensive than retaining an existing one.

The objective of this project is to:

- Analyze customer behavior
- Identify factors contributing to churn
- Build a machine learning model to predict churn
- Generate actionable business insights
- Develop an interactive dashboard for decision-makers

## Business Questions

1. What is the overall churn rate?
2. Which customer segments have the highest churn?
3. Which features influence churn the most?
4. Can churn be predicted accurately?
5. What recommendations can reduce churn?
